## Exercise #2:
Create a subset of data representing pokemon whose primary type is **Fire** 🔥 , and the second type is **Flying** 🦅. Save this DataFrame to the variable name `fire_flying_pokemon`. Display the first 5 lines of this subset.

```python
typeOne = poke['Type 1']
typeTwo = poke['Type 2']
fire_flying_pokemon = poke[(typeOne == 'Fire') & (typeTwo == 'Flying')]
fire_flying_pokemon.head()
```

## Exercise #3
Using `fire_flying_pokemon`, what is the median **Attack** of Fire Flying Pokemon?

```python
attack = fire_flying_pokemon['Attack'].median()
attack
```

## Exercise #5

Which primary type of pokemon is the overall strongest? (not looking for philsophical debates here, I'm looking at the average of the 'Total' column, grouped by Type-1 -  Fire v Water v Grass etc.).
- *Hint: Don't forget to actually answer the question here...*

```python
strongest = poke.groupby('Type 1')['Total'].mean()
sorted = strongest.sort_values(ascending=False)
sorted
# Dragon
```
